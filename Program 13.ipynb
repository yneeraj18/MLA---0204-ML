```python

import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    'Year': [2015, 2016, 2017, 2018, 2019,
             2020, 2021, 2022, 2023, 2024],
    'Kilometers': [80000, 70000, 60000, 50000, 40000,
                   35000, 30000, 20000, 15000, 10000],
    'Price': [300000, 350000, 400000, 450000, 500000,
              550000, 600000, 650000, 700000, 750000]
}

df = pd.DataFrame(data)

X = df[['Year', 'Kilometers']]
y = df['Price']

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[2022, 25000]])

print("Predicted Car Price:", prediction[0])
```
