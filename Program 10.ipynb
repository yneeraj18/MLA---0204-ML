```python

import numpy as np
from sklearn.mixture import GaussianMixture

X = np.array([
    [1, 2], [2, 3], [3, 3],
    [8, 8], [9, 8], [8, 9]
])

model = GaussianMixture(
    n_components=2,
    random_state=42
)

model.fit(X)

y_pred = model.predict(X)

print("Original Data:")
print(X)

print("Predicted Clusters:")
print(y_pred)

print("Cluster Means:")
print(model.means_)
```
